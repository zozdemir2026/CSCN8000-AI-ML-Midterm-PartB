# Problem 1 – Wine classification

##  Import Libraries

In this section, I imported the libraries needed to load the Wine dataset, prepare the data, train the models, and compare their accuracy.

```python
# Numerical operations and data manipulation
import numpy as np
import pandas as pd

# Load the Wine dataset
from sklearn.datasets import load_wine

# Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Naive Bayes classification models
from sklearn.naive_bayes import GaussianNB, MultinomialNB

# Feature preprocessing
from sklearn.preprocessing import MinMaxScaler

# Create machine learning pipelines
from sklearn.pipeline import make_pipeline

# Model evaluation
from sklearn.metrics import accuracy_score

print("All required libraries imported successfully!")
```

## Load Dataset and Basic EDA

In this section, I loaded the Wine dataset and performed basic exploratory data analysis (EDA).
I separated the Wine dataset, which contains 178 observations, into **X (independent variables)** and **y (dependent variable)**.

- **X: Features (Independent Variables):** Contains the 13 wine features used to predict the wine category.
- **y: Target (Dependent Variable):** Contains the actual wine categories (0, 1, and 2) that the model needs to predict.

I also checked the dataset size, feature names, wine categories, missing values, and the first five rows to understand the data before training the models.

```python
wine = load_wine()
X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = pd.Series(wine.target, name="class")
 
print("Dataset shape:", X.shape)
print("Feature names:", wine.feature_names)
print("Class names:", wine.target_names)
print("Examples per class:\n", y.value_counts().sort_index())
print("Missing values:", X.isna().sum().sum())
display(X.head())
display(X.describe())
```

## Split into training and test sets

The Wine dataset contains 178 observations. I split the dataset into 80% for training and 20% for testing.

- **Training data (80%):** Used to train the model to recognize wine categories.
- **Testing data (20%):** Used to evaluate how accurately the model predicts wine categories.

I used `random_state=42` to get the same data split every time I run the code.

I also used `stratify=y` to maintain a similar proportion of wine categories in both sets.

After splitting, I got 142 observations for training and 36 observations for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Train:", X_train.shape, "Test:", X_test.shape)
```

## Train GaussianNB

The GaussianNB model achieved an accuracy of **97.2%** on the testing data. This means the model correctly predicted 35 out of 36 wine observations. I learned that GaussianNB performs well on this dataset and can predict wine categories accurately using numerical features.

```python
gaussian_model = GaussianNB()
gaussian_model.fit(X_train, y_train)
gaussian_predictions = gaussian_model.predict(X_test)
gaussian_accuracy = accuracy_score(y_test, gaussian_predictions)
print(f"GaussianNB accuracy: {gaussian_accuracy:.3f}")
```

## Train MultinomialNB with non-negative features

I used Multinomial Naive Bayes to predict the three wine categories and compare its accuracy with GaussianNB.

MultinomialNB requires non-negative values, so I used MinMaxScaler to scale the features before training.

I trained the model using 142 observations and tested it using 36 observations.

```python
multinomial_model = make_pipeline(MinMaxScaler(clip=True), MultinomialNB())
multinomial_model.fit(X_train, y_train)
multinomial_predictions = multinomial_model.predict(X_test)
multinomial_accuracy = accuracy_score(y_test, multinomial_predictions)
print(f"MultinomialNB accuracy: {multinomial_accuracy:.3f}")
```

### Observation

I noticed that GaussianNB achieved better accuracy (97.2%) than MultinomialNB (94.4%). GaussianNB performed better because it may be more suitable for the continuous numerical features in the Wine dataset. I learned that different models can give different accuracy results even when using the same dataset.

##  Model Comparison and Predictions

In this section, I compared the accuracy of GaussianNB and MultinomialNB to find which model performed better.

I also created a table to compare the actual wine categories with the predictions from both models.

```python
print(f"GaussianNB:    {gaussian_accuracy:.3f}")
print(f"MultinomialNB: {multinomial_accuracy:.3f}")
if gaussian_accuracy > multinomial_accuracy:
    print("This is the best model : GaussianNB")
elif multinomial_accuracy > gaussian_accuracy:
    print("This is the our best naive  model: MultinomialNB")
else:
    print("Both models have equal test accuracy")
 
results = pd.DataFrame({
    "Actual": y_test.to_numpy(),
    "GaussianNB prediction": gaussian_predictions,
    "MultinomialNB prediction": multinomial_predictions
})
display(results.head(12))
```
