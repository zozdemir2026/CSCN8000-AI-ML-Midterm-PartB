## Importing Libraries

We import the necessary Python libraries for working with the Iris dataset. These libraries help with data processing, training the logistic regression model, and evaluating its performance using cross-validation.

```python
# Numerical operations and data manipulation
import numpy as np
import pandas as pd

# Load the Iris dataset
from sklearn.datasets import load_iris

# Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

# Naive Bayes classification models
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB, MultinomialNB

# Feature preprocessing
from sklearn.preprocessing import MinMaxScaler

# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Create machine learning pipelines
from sklearn.pipeline import make_pipeline

# Model evaluation
from sklearn.metrics import accuracy_score
from sklearn.model_selection import StratifiedKFold, cross_val_score

print("All required libraries imported successfully!")
```

## Load and inspect the Iris dataset by EDA 

In this section, I loaded the Iris dataset from Scikit-learn and assigned the features to **X_iris (independent variables)** and the target to **y_iris (dependent variable)**.

I performed basic EDA to explore and understand the dataset. I checked the DataFrame size, feature names, flower classes, and the number of observations in each class.

I also displayed the first five rows and used `describe()` to check basic statistics such as mean, minimum, and maximum values.

```python
iris = load_iris()
X_iris = pd.DataFrame(iris.data, columns=iris.feature_names)
y_iris = pd.Series(iris.target, name="species")
print("Shape:", X_iris.shape)
print("Features:", iris.feature_names)
print("Classes:", iris.target_names)
print("Count per class:\n", y_iris.value_counts().sort_index())
display(X_iris.head())
display(X_iris.describe())
```

## Train Logistic Regression Model

I used Logistic Regression to classify three Iris flower species: **Setosa, Versicolor, and Virginica**.

I used StandardScaler to keep the numerical features on a similar scale and help the model train better.

I also used a pipeline to combine scaling and model training. Then I trained the model using **X_iris (features)** and **y_iris (target)**.

```python
logistic_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)
logistic_model.fit(X_iris, y_iris)
print("Model fitted successfully.")
```

## Evaluate using five-fold stratified cross-validation

In this section, I used 5-fold cross-validation to check the Logistic Regression model's accuracy.

I divided the 150 Iris observations into 5 folds. Each time, the model used 120 observations for training and 30 for testing.

I used StratifiedKFold to keep the flower categories balanced in each fold.

I also calculated the mean accuracy and standard deviation to understand the model's performance.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(
    logistic_model, X_iris, y_iris, cv=cv, scoring="accuracy"
)
for fold, score in enumerate(scores, start=1):
    print(f"Fold {fold}: {score:.3f}")
print(f"Mean accuracy: {scores.mean():.3f}")
print(f"Standard deviation: {scores.std():.3f}")
```

### Observation

I noticed that the accuracy was different in each fold, ranging from 90% to 100%.

The mean accuracy was 95.3%, and the standard deviation was 0.045.

I learned that cross-validation helps us check how well the model performs on different parts of the dataset instead of depending on only one train/test split.
